```python
#challenge 61-70
import pandas as pd
import numpy as np

# Sample DataFrame with missing values
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [25, np.nan, 35, 40, np.nan],
    'Salary': [55000, 65000, np.nan, 85000, 95000],
    'Department': ['HR', 'Finance', 'IT', np.nan, 'Finance']
}
df = pd.DataFrame(data)

# Challenge 61: Find rows with missing values
print("\nChallenge 61:\n",df[df.isnull().any(axis=1)])

# Challenge 62: Fill missing Age with average age
df['Age'].fillna(df['Age'].mean(), inplace=True)
print("\nChallenge 62:\n", df)

# Challenge 63: Fill missing Salary with median salary
# Challenge 63: Drop rows where Salary is missing
df = df.dropna(subset=['Salary'])
print("\nChallenge 63:\n", df)

# Challenge 64: Fill missing Department with 'Unknown'
df['Department'].fillna('Unknown', inplace=True)
print("\nChallenge 64:\n", df)

# Challenge 65: Drop rows where Age is missing
df = df.dropna(subset=['Age'])
print("\nChallenge 65:\n", df)

# Challenge 66: Merge df with df2 on 'Department'
merged = pd.merge(df, df2, on='Department', how='left')
print("\nChallenge 66:\n", merged)

# Challenge 67: Create a pivot table showing average salary per Department
pivot = pd.pivot_table(df, index='Department', values='Salary', aggfunc='mean')
print("\nChallenge 67:\n", pivot)

# Challenge 68: Map departments to short codes
dept_map = {'HR': 'H', 'Finance': 'F', 'IT': 'I', 'Marketing': 'M', 'Unknown': 'U'}
df['Dept_Code'] = df['Department'].map(dept_map)
print("\nChallenge 68:\n", df)

# Challenge 69: Bin ages into groups (e.g., <30 = 'Young', 30-40 = 'Mid', >40 = 'Senior')
df['Age_Group'] = pd.cut(df['Age'], bins=[0, 30, 40, 100], labels=['Young', 'Mid', 'Senior'])
print("\nChallenge 69:\n", df)

# Challenge 70: Create a crosstab between Age_Group and Department
crosstab = pd.crosstab(df['Age_Group'], df['Department'])
print("\nChallenge 70:\n", crosstab)

```

```python
#Challenge 71-80
# Sample data
data = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Department': ['HR', 'Finance', 'IT', 'HR', 'Finance'],
    'Performance_Score': [88, 92, 95, 70, 85],
    'Salary': [55000, 65000, 75000, 85000, 95000]
}
df = pd.DataFrame(data)

# Challenge 71: Set 'Employee_ID' as index
df.set_index('Employee_ID', inplace=True)
print("\nChallenge 71:\n", df)

# Challenge 72: Reset the index
df.reset_index(inplace=True)
print("\nChallenge 72:\n", df)

# Challenge 73: Rename column 'Salary' to 'CTC'
df.rename(columns={'Salary': 'CTC'}, inplace=True)
print("\nChallenge 73:\n", df)

# Challenge 74: Apply 10% hike to all CTCs
df['CTC_Hike'] = df['CTC'] * 1.10
print("\nChallenge 74:\n", df)

# Challenge 75: Add a Boolean column 'High_Performer' where score > 90
df['High_Performer'] = df['Performance_Score'] > 90
print("\nChallenge 75:\n", df)

# Challenge 76: Count of high performers by Department
print("\nChallenge 76:\n", df[df['High_Performer']].groupby('Department')['High_Performer'].count())

# Challenge 77: Use `apply()` to label performance: <80 = 'Low', 80-90 = 'Avg', >90 = 'High'
def performance_label(score):
    if score < 80:
        return 'Low'
    elif score <= 90:
        return 'Avg'
    else:
        return 'High'


df['Perf_Label'] = df['Performance_Score'].apply(performance_label)
print("\nChallenge 77:\n", df)

# Challenge 78: Display top 3 employees with highest CTC
print("\nChallenge 78:\n", df.nlargest(3, 'CTC'))

# Challenge 79: Groupby Department and get average Performance Score
print("\nChallenge 79:\n", df.groupby('Department')['Performance_Score'].mean())

# Challenge 80: Save the final dataframe to CSV
df.to_csv('employee_summary.csv', index=False)
print("\nChallenge 80:\nCSV 'employee_summary.csv' saved successfully!")
```
